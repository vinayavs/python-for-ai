Syntax is about the correct arrangement of words and symbols in a code.

Semantics refers to the meaning or the interpretation of the symbols, characters, and commands in a language. It is about what the code is supposed to do when it runs.

```python
## Basic Syntax Rules In Python
## Case sensitivity- Python is case sensitive

a = 10
A = 20
print(a)  # Output: 10
print(A)  # Output: 20
```

```python
# Indentation- Python uses indentation to define blocks of code. Indentation is crucial for defining the structure of your code.
test = 5
if test > 0:
    print("Positive number")
```

```python
# Line continuation- Python allows you to break long lines of code into multiple lines using the backslash (\) character or by using parentheses.
# Using backslash
result = 1 + 2 + 3 + \
         4 + 5
print(result)  # Output: 15
```

```python
# Multi-line statements- You can use parentheses to create multi-line statements without using backslashes.
total = (1 + 2 + 3 +
         4 + 5)
print(total)  # Output: 15
```

```python
# Multiple statements on a single line- You can write multiple statements on a single line using semicolons (;).
x = 1; y = 2; z = 3
print(x, y, z)  # Output: 1 2 3
```

```python
# Variable Assignment- In Python, you can assign values to multiple variables in a single line.
a, b, c = 1, 2, 3
d="Hello"
print(a, b, c)  # Output: 1 2 3
print(type(a))  # Output: <class 'int'>
print(type(d))  # Output: <class 'str'>
```

```python
# Type Inference- Python automatically infers the type of a variable based on the value assigned to it. You don't need to explicitly declare the type of a variable.
x = 5
print(type(x))  # Output: <class 'int'>
y = 3.14
y="Hi"
print(type(y))  # Output: <class 'str'>
```
