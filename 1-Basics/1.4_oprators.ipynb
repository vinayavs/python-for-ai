OPERATORS IN PYTHON
1. Arithmetic Operators :
    Addition,
    Subtraction,
    Multiplication,
    Division,
    Floor Division,
    Modulus,
    Exponentiation.
2. Comparison Operators
    Equal to, 
    Not equal to,
    Greater than,
    Less than,
    Greater than or equal to,
    Less than or equal to.
3. Logical Operators
    AND,
    OR,
    NOT.

```python
# Operators in Python are special symbols that carry out arithmetic or logical computation or comparison. 
# The value that the operator operates on is called the operand. For example: in 2 + 3, + is the operator and 2 and 3 are the operands.
a=10
b=5

# Arithmetic Operators
a+b  # Addition: Output: 15
a-b  # Subtraction: Output: 5
a*b  # Multiplication: Output: 50
a/b  # Division: Output: 2.0
a//b  # Floor Division: Output: 2
a%b  # Modulus: Output: 0
a**b  # Exponentiation: Output: 100000  

# Comparison Operators
a==b  # Equal to: Output: False
a!=b  # Not equal to: Output: True
a>b  # Greater than: Output: True
a<b  # Less than: Output: False 
a>=b  # Greater than or equal to: Output: True
a<=b  # Less than or equal to: Output: False

# Logical Operators
a>5 and b<10  # Logical AND: Output: True
a>5 or b<10  # Logical OR: Output: True
not(a>5)  # Logical NOT: Output: False 

x= True
y= False
x and y  # Logical AND: Output: False
x or y  # Logical OR: Output: True
not x  # Logical NOT: Output: False
not y  # Logical NOT: Output: True

True and True  # Output: True
True and False  # Output: False
False and True  # Output: False
False and False  # Output: False

True or False  # Output: True 
False or True  # Output: True
True or True  # Output: True
False or False  # Output: False
```
