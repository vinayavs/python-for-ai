Variables In Python

```python
# Variable in python is a reserved memory location to store values. In other words, a variable in a python program gives data to the computer for processing. 
# They do not need explicit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable. The equal sign (=) is used to assign values to variables.
a=10
```

```python
# Declaring and assigning a value to a variable
# Variables and Data Types- Python is dynamically typed, meaning you don't need to declare the type of a variable explicitly. The type is inferred from the value assigned to it.
a = 10
b = "test"
c = 3.14
d = True
e = [1, 2, 3]
f = (1, 2, 3)
g = {'name': 'Alice', 'age': 25}
h = None
i = 1 + 2j
j = {1, 2, 3}
k = frozenset([1, 2, 3])
l = bytearray(b"Hello")
m = bytes(b"Hello")
n = memoryview(b"Hello")
o = range(5)
p = complex(2, 3)
q = lambda x: x + 1
r = type(a), type(b), type(c), type(d), type(e), type(f)
type(a), type(b), type(c), type(d), type(e), type(f), type(g), type(h), type(i), type(j), type(k), type(l), type(m), type(n), type(o), type(p), type(q), type(r)
```

```python
# Naming conventions for variables in Python:
# 1. Variable names can contain letters, numbers, and underscores (_), but they cannot start with a number.
# 2. Variable names are case-sensitive, meaning that 'myVariable' and 'myvariable' would be considered different variables.
# 3. Variable names should be descriptive and meaningful, reflecting the purpose of the variable.
# 4. Avoid using Python reserved keywords 
a = 10
A = 20
print(a)  # Output: 10
print(A)  # Output: 20
print(a==A)  # Output: False
_a = 30
print(_a)  # Output: 30
# 1abc = 40  # Invalid variable name, cannot start with a number
```

```python
# Type Checking and Type Conversion:
# Python provides built-in functions to check the type of a variable and convert between different data types
x = 10
y = "20"
print(type(x))  # Output: <class 'int'>
print(type(y))  # Output: <class 'str'>
# Type conversion
x_str = str(x)  # Convert integer to string
y_int = int(y)  # Convert string to integer
print(type(x_str))  # Output: <class 'str'>
print(type(y_int))  # Output: <class 'int'>

b = 3.14
b_int = int(b)  # Convert float to integer
print(b_int)  # Output: 3
print(type(b_int))  # Output: <class 'int'>
float_str = float(b_int)  # Convert float to string
print(float_str)  # Output: 3.0
print(type(float_str))  # Output: <class 'float'>

a = "test"
a_int = int(a)  # Convert to integer
print(type(a_int))  # Output: <class 'int'>
```

```python
# Input
a = input("Enter a number: ")
print("You entered:", a, type(a))  # Output: You entered: <input_value> <class 'str'>

b = int(input("Enter another number: "))
print("You entered:", b, type(b))  # Output: You entered: <input_value> <class 'int'>
```
