CONDITIONAL STATEMENTS IN PYTHON

```python
# Conditional Statements in Python
# if statement else statement elif statement nested if statement
# if statement
x = 10
if x > 5:
    print("x is greater than 5")


# else statement
y = 3
if y > 5:
    print("y is greater than 5")
else:
    print("y is not greater than 5")


# elif statement
age = 18
if age >= 18:
    print("age is 18 or greater")
elif age >= 10:
    print("age is between 10 and 17")
else:
    print("age is less than 10")

```

```python
# nested if statement
num = int(input("Enter a number: "))
if num > 0:
    print("num is positive")
    if num%2 == 0:
        print("num is even")
    else:
        print("num is odd")
```

```python
# Examples.
# Leap year or not
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
# nested if
year = int(input("Enter a year: "))
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year" )
```

```python
# Simple Calc
operation = input("Enter the operation :") # + - / %
x = float(input("Enter 1st value to calculate :"))
y = float(input("Enter 2nd value to calculate :"))
res = 0.0
if operation == "+":
    res = x + y
    print("Result is: ",res)
elif operation == "-":
    res = x - y 
    print("Result is: ",res)
elif operation == "*":
    res = x * y
    print("Result is: ",res)
elif operation == "/":
    if res != 0:
        res = x / y 
        print("Result is: ",res)
    else:
        print("Error!! Division by zero")
elif operation == "%":
    res = x % y
    print("Result is: ",res)
else:
    print("Invalid operation")
```
