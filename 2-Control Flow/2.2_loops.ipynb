LOOPS IN PYTHON

```python
# Loops in python are used to execute a block of code repeatedly until a certain condition is met. 
# There are two main types of loops in Python: the `for` loop and the `while` loop.
# Loop control statements are used to change the flow of execution in a loop. The most common loop control statements are `break`, `continue`, and `pass`.
# Nested loops are loops within loops, allowing for more complex iterations.
```

```python
# For Loop:
print("Range is :", range(5))
for i in range(5):
    print(i)
```

```python
for x in range(1,6):
    print(x)
```

```python
for i in range(2,20,2):
    print(i)
```

```python
for i in range(20,2,-2):
    print(i)
```

```python
# Stings 
str = "Hello World"
for i in str:
    print(i)
```

```python
# While loop: While loops continue to execute a block of code as long as a specified condition is true.
count = 0 
while count < 5:
    print(count)
    count = count + 1
```

```python
# Loop Control Statements:
# Break Statement: The `break` statement is used to exit a loop prematurely when a certain condition is met.
for i in range(10):
    if i == 5:
        print("Breaking the loop at i =", i)
        break
        print("test")  # This line will not be executed because of the break statement
    print(i)
```

```python
# Continue Statement: The `continue` statement is used to skip the current iteration of a loop and move on to the next iteration.
for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
# Pass Statement: The `pass` statement is a null operation; it is used as a placeholder in loops or functions where code is syntactically required but no action is needed.
for i in range(5):
    if i == 3:
        print("Passing the loop at i =",i)
        pass
    print(i)
```

```python
# Nested Loops: Nested loops are loops within loops, allowing for more complex iterations.
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
# Cal sum of first N natural numbers using while and for loops
# While loop

N = 10
sum = 0
count = 1
while count <= N:
    sum = sum + count
    count = count + 1
print(f"Total is using while loop:{sum}") 

# For loop
sum = 0
for i in range(11):
    sum = sum + i
print(f"Total is using for loop:{sum} ")
```

```python
# Get prime numbers between 1 to 100
for i in range(1, 101):
    if i>1:
        for j in range(2,i):
            if i%j==0:
                break
        else:
            print(i, "is a prime number")   

```

```python
items = ["apple", "oragnge", "cherry"]
search_target = "blueberry"

for item in items:
    if item == search_target:
        print(f"Found {search_target}!")
        break  # Exits the loop, skipping the else block
else:
    # This runs because the loop finished without hitting 'break'
    print(f"Could not find {search_target} in the list.")
```

```python
# Get prime numbers between 1 to 10
for i in range(1, 11):
    if i>1:
        for j in range(2,i):
            print(f"i:{i} and j:{j}")
            print(f"Checking {i%j}")
            if i%j==0:
                print(f"{i} is not a prime number")
                print("Breaking the loop at j =", j)
                break
        else:
            print(f"{i} is a prime number") 
```
