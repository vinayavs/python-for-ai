LOOPS IN PYTHON

```python
# Loops in python are used to execute a block of code repeatedly until a certain condition is met. 
# There are two main types of loops in Python: the `for` loop and the `while` loop.
# Loop control statements are used to change the flow of execution in a loop. The most common loop control statements are `break`, `continue`, and `pass`.
# Nested loops are loops within loops, allowing for more complex iterations.
```

```python
# For Loop:
print("Range is :", range(5))
for i in range(5):
    print(i)
```

```python
for x in range(1,6):
    print(x)
```

```python
for i in range(2,20,2):
    print(i)
```

```python
for i in range(20,2,-2):
    print(i)
```

```python
# Stings 
str = "Hello World"
for i in str:
    print(i)
```
